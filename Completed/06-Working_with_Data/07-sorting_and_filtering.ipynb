# 🧮 Sorting & Filtering in Pandas

## 🔹 LEARNING GOALS:
- Sort rows using `.sort_values()`
- Filter data using conditional logic
- Chain multiple filters together
- Use `.query()` for readable filtering


# The "students.csv" file coundn't be found.
# I am using the "mock_students.csv" file that I generated using Mockaroo for this notebook.

### 📥 1. Load the Dataset

```python
# Load the "mock_students.csv" file
from google.colab import files
uploaded = files.upload()
```

```python
import pandas as pd

df = pd.read_csv("mock_students.csv")
df.head()
```

### 🔢 2. Sorting Data

```python
# Sort by math_score
df.sort_values(by="math_score")
```

```python
# Sort by english_score descending
df.sort_values(by="english_score", ascending=False)
```

```python
# Sort by multiple columns
df.sort_values(by=["english_score", "math_score"], ascending=[False, True])
```

### 🧪 3. Filtering with Conditions

```python
# Who scored at least 90 in math?
df[df["math_score"] >= 90]
```

```python
# Who scored less than 70 in either subject?
df[(df["math_score"] < 70) | (df["english_score"] < 70)]
```

### 🔗 4. Combining Filters

```python
# Students with B+ or higher in both
df[(df["math_score"] >= 87) & (df["english_score"] >= 87)]
```

### 🧾 5. Using `.query()` for Readability

```python
# Same query using .query()
df.query("math_score >= 87 and english_score >= 87")
```

### 🧪 Try It Yourself

- Sort the dataset by first name alphabetically
- Find students with the same math and science score
- Use `.query()` to find students who scored over 95 in *either* subject


```python
# Sort by first name alphabetically
df.sort_values(by="first_name", ascending=True)
```

```python
# Find students with same Math and English score
df[df["math_score"] == df["english_score"]]
```

```python
# Use .query() to find students scoring over 95 in either subject
df.query("math_score > 95 or english_score > 95")
```
