## Q4. File I/O
Write Python code that saves `Hello World` to a file called `hello.txt`, then reads it back and prints the contents.

```python
# Write to a file
with open("hello.txt", "w") as f:
    f.write("Hellow World")

# Reading the file back
with open("hello.txt", "r") as f:
    contents = f.read()

# Print the contents
contents
```

## Q6. Plotting
Create a line plot of $y = x^2$ for $x$ values from 0 to 10.

```python
import numpy as np
import matplotlib.pyplot as plt

# x values from 0 to 10
x = np.linspace(0, 10, 200)

y = x**2

plt.title("Line Plot of y = x^2")
plt.plot(x, y, label="y = x^2")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

## Q9. Timing
Compare the runtime of squaring numbers using a Python loop vs. NumPy. Use `%timeit` for each.

```python
import numpy as np

numsList = list(range(1000))
nums_np = np.arange(1000)

# Python loop
%timeit [x**2 for x in numsList]

# NumPy
%timeit nums_np**2
```
